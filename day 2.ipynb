Argument types in python

```python
# positional arguments
def describe(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}")
```

```python
describe('dog','buddy')   # order matters : animal type and pet name
```

```python
describe('buddy','dog')
```

```python
# Default arguments
def greet_user(name, greeting='Hello'):
    print(f'{greeting}, {name}')
```

```python
greet_user('Bob')   # uses default : hello bob
```

```python
greet_user("Bob","Good Morning")
```

```python
# ketword arguments
describe(pet_name='buddy',animal_type='dog')
```

```python
# Arbitrary positional arguments(*args)
def print_fruits(*fruits):
    for fruit in fruits:
        print(fruit)
```

```python
print_fruits("Apple", "Banana", "Cherry")
```

```python
# Arbitrary keyword arguments (**kwargs)
def show_details(**info):
    for key, value in info.items():
        print(f"{key} : {value}")
```

```python
show_details(name='alice',age=25, city='kathmandu')
```

### Filter function

```python
def is_even(num):
    return num % 2 == 0 
```

```python
num = [1,2,3,4,5,6]
even_num = filter(is_even,num)
```

```python
for i in even_num:
    print(i)
```

```python
mixed_data = [True, False, "", 0,42,None, "Hello"]
clean_data = filter(None, mixed_data)
```

```python
list(clean_data)
```

### Any function

```python
def any(iterable):
    for element in iterable:
        if element:
            return True
    return False
```

```python
any([1,2,3,4])
```

```python
any([])
```

```python
any(["",4])
```
