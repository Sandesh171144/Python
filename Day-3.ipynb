```python
class Dog:
    species = 'Canine'  # class variable

    def __init__(self,name,breed):
        # instance variable
        self.name = name
        self.breed = breed
```

```python
class Dog:
    def __init__(self,name):
        self.name = name

    def __str__(self):
        return f"Dog named {self.name}"
        
```

```python
class Calculator:
    def __init__(self,brand):
        self.brand = brand 

    def get_brand(self):
        return self.brand

    @classmethod
    def info(cls):
        return f" This is a {cls.__name__} class."

    @staticmethod
    def add(x,y):
        return(x + y)
```

```python
cal = Calculator('Casio')
```

```python
cal.get_brand()
```

```python
cal.info()
```

```python
cal.add(3,4)
```
